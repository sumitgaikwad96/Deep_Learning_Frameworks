# Deep Learning Frameworks



## 1. What is TensorFlow 2.0, and how is it different from TensorFlow 1.x?

**Answer:** TensorFlow 2.0 is Google's open-source machine-learning framework. Compared with TensorFlow 1.x, it uses eager execution by default, provides the integrated tf.keras high-level API, and uses tf.function when graph execution and optimization are needed.

## 2. How do you install TensorFlow 2.0?

**Answer:** Use pip: `pip install tensorflow`. For a specific compatible release, a version specifier can be used.

## 3. What is the primary function of tf.function in TensorFlow 2.0?

**Answer:** `tf.function` converts a Python function into a TensorFlow graph function, enabling graph-based optimization and potentially faster execution.

## 4. What is the purpose of the Model class in TensorFlow 2.0?

**Answer:** `tf.keras.Model` represents a trainable neural-network model. It groups layers and supports compilation, training, evaluation, prediction, and saving.

## 5. How do you create a neural network using TensorFlow 2.0?

**Answer:** Use `tf.keras.Sequential` or subclass `tf.keras.Model`, add layers, then call `compile()` and `fit()` to configure and train the network.

## 6. What is the importance of TensorSpace in TensorFlow?

**Answer:** TensorSpace is a visualization tool for exploring neural-network architecture and layer computations, helping users understand models visually.

## 7. How can TensorBoard be integrated with TensorFlow 2.0?

**Answer:** Use the Keras `TensorBoard` callback, for example `tf.keras.callbacks.TensorBoard(log_dir='logs')`, while training with `model.fit()`.

## 8. What is the purpose of TensorFlow Playground?

**Answer:** It is an interactive browser-based environment for experimenting with neural networks and observing how architecture and training settings affect learning.

## 9. What is Netron, and how is it useful for deep learning models?

**Answer:** Netron is a model-visualization tool. It helps inspect neural-network graphs, layers, operators, inputs, outputs, and architecture.

## 10. What is the difference between TensorFlow and PyTorch?

**Answer:** Both are deep-learning frameworks with tensors, automatic differentiation, GPU support, and neural-network APIs. TensorFlow has a broad ecosystem including Keras and deployment tooling; PyTorch is known for its Python-friendly dynamic computation style and torch.nn ecosystem.

## 11. How do you install PyTorch?

**Answer:** Install PyTorch using the installation command appropriate for the operating system, Python version, and CPU/GPU configuration. The exact command depends on the environment.

## 12. What is the basic structure of a PyTorch neural network?

**Answer:** Subclass `torch.nn.Module`, define layers in `__init__()`, and define the forward computation in `forward()`. Training uses a forward pass, loss, backpropagation, and optimizer update.

## 13. What is the significance of tensors in PyTorch?

**Answer:** Tensors are PyTorch's core multidimensional data structure. They store inputs, parameters, activations, and gradients and can run on CPU or GPU.

## 14. What is the difference between torch.Tensor and torch.cuda.Tensor in PyTorch?

**Answer:** A CPU tensor normally resides on the CPU. A CUDA tensor resides on a CUDA-capable GPU. Modern PyTorch commonly controls this using `.to(device)`.

## 15. What is the purpose of the torch.optim module in PyTorch?

**Answer:** `torch.optim` provides optimization algorithms such as SGD and Adam that update model parameters using computed gradients.

## 16. What are some common activation functions used in neural networks?

**Answer:** Common functions include ReLU, sigmoid, tanh, softmax, Leaky ReLU, and GELU.

## 17. What is the difference between torch.nn.Module and torch.nn.Sequential in PyTorch?

**Answer:** `nn.Module` is the base class for custom models and components with arbitrary forward logic. `nn.Sequential` is a convenience container that applies layers in a fixed sequence.

## 18. How can you monitor training progress in TensorFlow 2.0?

**Answer:** Use the `History` object returned by `model.fit()`, validation metrics, Matplotlib plots, TensorBoard, and callbacks such as EarlyStopping.

## 19. How does the Keras API fit into TensorFlow 2.0?

**Answer:** `tf.keras` is TensorFlow's high-level Keras API. It simplifies model construction, compilation, training, evaluation, prediction, and saving.

## 20. What is an example of a deep learning project that can be implemented using TensorFlow 2.0?

**Answer:** Examples include image classification with a CNN, object detection, text classification, regression, and time-series forecasting.

## 21. What is the main advantage of using pre-trained models in TensorFlow and PyTorch?

**Answer:** Pre-trained models provide learned representations from large datasets, reducing training time and data/computational requirements. They can be adapted using transfer learning or fine-tuning.

## 22. How do you install and verify that TensorFlow 2.0 was installed successfully?

**Answer:** Install TensorFlow and run `import tensorflow as tf; print(tf.__version__)`. Successful import and a 2.x version confirm the installation.

## 23. How can you define a simple function in TensorFlow 2.0 to perform addition?

**Answer:** Use `def add(a,b): return tf.add(a,b)`. Calling it with TensorFlow constants returns a TensorFlow tensor containing the sum.

## 24. How can you create a simple neural network in TensorFlow 2.0 with one hidden layer?

**Answer:** Create a `tf.keras.Sequential` model with an input layer, one Dense hidden layer using an activation such as ReLU, and an output Dense layer. Compile and train it.

## 25. How can you visualize the training progress using TensorFlow and Matplotlib?

**Answer:** Store the result of `model.fit()` in `history`, then plot values such as `history.history['loss']` and validation loss using Matplotlib.

## 26. How do you install PyTorch and verify the PyTorch installation?

**Answer:** Install PyTorch for the target environment, then run `import torch; print(torch.__version__)`. `torch.cuda.is_available()` can also check CUDA availability.

## 27. How do you create a simple neural network in PyTorch?

**Answer:** Subclass `nn.Module`, define `nn.Linear` layers in `__init__`, and connect them in `forward()` with an activation function.

## 28. How do you define a loss function and optimizer in PyTorch?

**Answer:** Choose a loss such as `nn.MSELoss()` or `nn.CrossEntropyLoss()`, then create an optimizer such as `torch.optim.Adam(model.parameters(), lr=0.001)`.

## 29. How do you implement a custom loss function in PyTorch?

**Answer:** Define a Python function using tensor operations that receives predictions and targets and returns a scalar loss tensor. It can then be used in the training loop.

## 30. How do you save and load a TensorFlow model?

**Answer:** Save with `model.save('model.keras')` and load with `tf.keras.models.load_model('model.keras')`.

## Practical 3. TensorFlow neural network with one hidden layer

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(2,)),
    tf.keras.layers.Dense(8, activation="relu"),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer="adam", loss="mse", metrics=["mae"])
model.summary()
```

## Practical 4. Visualize TensorFlow training progress using Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

x = np.arange(0, 10, dtype=np.float32).reshape(-1, 1)
y = 2 * x + 1

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(8, activation="relu"),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer="adam", loss="mse")
history = model.fit(x, y, epochs=50, verbose=0)

plt.figure(figsize=(8, 5))
plt.plot(history.history["loss"])
plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```

## Practical 6. Create a simple PyTorch neural network

```python
import torch
import torch.nn as nn

class SimpleNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(2, 8),
            nn.ReLU(),
            nn.Linear(8, 1)
        )

    def forward(self, x):
        return self.network(x)

model = SimpleNet()
print(model)
```

## Practical 7. Define a PyTorch loss function and optimizer

```python
import torch
import torch.nn as nn
import torch.optim as optim

model = SimpleNet()
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

x = torch.tensor([[1.0, 2.0], [2.0, 3.0]])
y = torch.tensor([[5.0], [7.0]])

prediction = model(x)
loss = criterion(prediction, y)

optimizer.zero_grad()
loss.backward()
optimizer.step()

print("Loss:", loss.item())
```

## Practical 8. Implement a custom PyTorch loss function

```python
import torch

def custom_mse_loss(prediction, target):
    return torch.mean((prediction - target) ** 2)

prediction = torch.tensor([2.0, 4.0, 6.0])
target = torch.tensor([1.0, 5.0, 7.0])

loss = custom_mse_loss(prediction, target)
print("Custom loss:", loss.item())
```

## Practical 9. Save and load a TensorFlow model

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(4, activation="relu"),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer="adam", loss="mse")

model.save("example_model.keras")
loaded_model = tf.keras.models.load_model("example_model.keras")

print("Model saved and loaded successfully.")
loaded_model.summary()
```
